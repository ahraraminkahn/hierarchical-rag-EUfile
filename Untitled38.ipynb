```python
from pypdf import PdfReader
#here we extract content and page no from each page by using a for loop and store it in the pages empty list
#which we would further use to build/enrich our metadata for better precise retrieval
PDF_PATH = "./EU-AI-Act.pdf"

reader = PdfReader(PDF_PATH, strict=False)

pages = []

for page_number, page in enumerate(reader.pages, start=1):
    try:
        text = page.extract_text()

        if text:
            pages.append({
                "page": page_number,
                "text": text
            })

    except Exception as e:
        print(f"Error on page {page_number}: {e}")

print("Pages extracted:", len(pages))
```

```python
#this cell separates each heading, sub-heading from the pages, to identify the structure of the hierarchy of the pdf
import re

chapter_pattern = re.compile(
    r"^(\d+)\s+(.+)$"
)

section_pattern = re.compile(
    r"^(\d+\.\d+)\s+(.+)$"
)

subsection_pattern = re.compile(
    r"^(\d+\.\d+\.\d+)\s+(.+)$"
)
```

```python
#Program ko yaad rahega ke woh abhi kis chapter/section/subsection mein hai.
current_chapter = None
current_chapter_title = None

current_section = None
current_section_title = None

current_subsection = None
current_subsection_title = None
```

```python
#Yeh cell PDF ki headings se hierarchy samajhne ki koshish kar raha hai.
structured_pages = []

for page in pages:

    page_text = page["text"]

    lines = page_text.split("\n")

    for line in lines:

        line = line.strip()

        if not line:
            continue

        # Subsection
        match = subsection_pattern.match(line)

        if match:
            current_subsection = match.group(1)
            current_subsection_title = match.group(2)

            continue

        # Section
        match = section_pattern.match(line)

        if match:
            current_section = match.group(1)
            current_section_title = match.group(2)

            current_subsection = None
            current_subsection_title = None

            continue

        # Chapter
        match = chapter_pattern.match(line)

        if match:
            current_chapter = match.group(1)
            current_chapter_title = match.group(2)

            current_section = None
            current_section_title = None

            current_subsection = None
            current_subsection_title = None

            continue
```

```python
#LangChain Documents banarahe hain
from langchain_core.documents import Document

documents = []

current_chapter = None
current_chapter_title = None

current_section = None
current_section_title = None

current_subsection = None
current_subsection_title = None


for page in pages:

    lines = page["text"].split("\n")

    page_content = []

    for line in lines:

        line = line.strip()

        if not line:
            continue

        # subsection
        match = subsection_pattern.match(line)

        if match:
            current_subsection = match.group(1)
            current_subsection_title = match.group(2)
            continue

        # section
        match = section_pattern.match(line)

        if match:
            current_section = match.group(1)
            current_section_title = match.group(2)

            current_subsection = None
            current_subsection_title = None

            continue

        # chapter
        match = chapter_pattern.match(line)

        if match:
            current_chapter = match.group(1)
            current_chapter_title = match.group(2)

            current_section = None
            current_section_title = None

            current_subsection = None
            current_subsection_title = None

            continue

        page_content.append(line)

    documents.append(
        Document(
            page_content="\n".join(page_content),
            metadata={
                "page": page["page"],
                "chapter": current_chapter,
                "chapter_title": current_chapter_title,
                "section": current_section,
                "section_title": current_section_title,
                "subsection": current_subsection,
                "subsection_title": current_subsection_title
            }
        )
    )
```

```python
#splitting the text using recursive text splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
    separators=[
        "\n\n",
        "\n",
        ". ",
        " ",
        ""
    ]
)

chunks = splitter.split_documents(documents)

print("Total chunks:", len(chunks))
```

```python
#add metadata to each chunk our text splitter has created above
for i, chunk in enumerate(chunks):

    chunk.metadata["chunk_id"] = f"chunk_{i}"

    chunk.metadata["chunk_index"] = i

    chunk.metadata["document"] = "Understanding Deep Learning"

    chunk.metadata["author"] = "Simon J. D. Prince"

    chunk.metadata["content_type"] = "text"
```

```python
#Content type function:
#Chunk ke content ko category assign kar raha hai.
def detect_content_type(text):

    text_lower = text.lower()

    if "definition" in text_lower:
        return "definition"

    if "example" in text_lower:
        return "example"

    if "theorem" in text_lower:
        return "theorem"

    if "algorithm" in text_lower:
        return "algorithm"

    if "=" in text:
        return "equation_or_math"

    return "concept"
```

```python
#Chunk ka text function ko diya aur result metadata mein save.
for chunk in chunks:

    chunk.metadata["content_type"] = detect_content_type(
        chunk.page_content
    )
```

```python
#Agar subsection available hai, woh topic banega.
for chunk in chunks:

    chunk.metadata["topic"] = (
        chunk.metadata.get("subsection_title")
        or chunk.metadata.get("section_title")
        or chunk.metadata.get("chapter_title")
    )
```

```python
#ID se chunk ki hierarchy bhi pata chal sakti hai.
for i, chunk in enumerate(chunks):

    chapter = chunk.metadata.get("chapter") or "unknown"
    section = chunk.metadata.get("section") or "unknown"
    subsection = chunk.metadata.get("subsection") or "unknown"

    chunk.metadata["chunk_id"] = (
        f"ch{chapter}_"
        f"sec{section}_"
        f"sub{subsection}_"
        f"chunk{i}"
    )
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
#used to setup qdrant
from langchain_qdrant import QdrantVectorStore
from qdrant_client import QdrantClient
from qdrant_client.http.models import Distance, VectorParams

#collection name
COLLECTION_NAME = "sample"

EMBED_DIM = len(embeddings.embed_query("dimension probe"))

client = QdrantClient(path="qdrant_db")

if not client.collection_exists(COLLECTION_NAME):
    client.create_collection(
        collection_name=COLLECTION_NAME,

        vectors_config=VectorParams(size=EMBED_DIM, distance=Distance.COSINE),
    )

vector_store = QdrantVectorStore(
    client=client,
    collection_name=COLLECTION_NAME,
    embedding=embeddings,
)
vector_store.add_documents(chunks)

print("Number of stored chunks:", client.count(COLLECTION_NAME).count)
```

```python
from langchain_community.retrievers import BM25Retriever
from langchain_classic.retrievers import EnsembleRetriever

vector_retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 10},
)

bm25_retriever = BM25Retriever.from_documents(chunks)
bm25_retriever.k = 10

base_retriever = EnsembleRetriever(
    retrievers=[vector_retriever, bm25_retriever],
    weights=[0.5, 0.5],
)
```

```python
from langchain_community.cross_encoders import HuggingFaceCrossEncoder

from langchain_classic.retrievers import ContextualCompressionRetriever
from langchain_classic.retrievers.document_compressors import CrossEncoderReranker

cross_encoder = HuggingFaceCrossEncoder(model_name="cross-encoder/ms-marco-MiniLM-L-6-v2")

reranker = CrossEncoderReranker(model=cross_encoder, top_n=3)

retriever = ContextualCompressionRetriever(
    base_compressor=reranker,
    base_retriever=base_retriever,
)
```

```python
import os
from langchain_openai import ChatOpenAI
from google.colab import userdata

llm = ChatOpenAI(
    model="gpt-4o-mini",
    openai_api_key = userdata.get("OPENAI_API_KEY"),
)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableLambda, RunnablePassthrough

def get_text(documents):
    return "\n\n".join(doc.page_content for doc in documents)


prompt = ChatPromptTemplate.from_template(
    "You are answering a question using only the context provided below. "
    "If the answer isn't contained in the context, say you don't know instead of guessing.\n\n"
    "Context:\n{context}\n\n"
    "Question: {question}\n\n"
    "Answer:"
)

chain = (
    {"context": retriever | RunnableLambda(get_text), "question": RunnablePassthrough()}
    | prompt
    | llm
)
```

```python
query = "what are Classification rules for high-risk AI systems "

response = chain.invoke(query)
print(response.content)
```
